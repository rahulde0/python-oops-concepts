Encapsulation =
✔ bundling data + methods together
✔ controlling access using public, protected, private conventions

Python uses naming conventions:

var → public

_var → protected

__var → private (name-mangled)

```python
# ============================================================
# CONCEPT 3 — ENCAPSULATION IN PYTHON
# ============================================================
# Encapsulation means:
# 1. Hiding internal data from outside access.
# 2. Providing controlled access through methods (getters/setters).
# 3. Grouping related data + behavior inside classes.
# ============================================================


# ------------------------------------------------------------
# Example 1 — Public attribute (fully accessible)
# ------------------------------------------------------------
class Person:
    surname= 'deo'
    def __init__(self):
        self.name = "Rahul"
        self.surname="DDDeo" # public variable

p = Person()
print(p.name)
print(p.surname)
print(Person.surname)

# Explanation:
# Public variables can be accessed directly.
# No restrictions apply — default accessibility in Python.

```

```python

# ------------------------------------------------------------
# Example 2 — Protected attribute (_var)
# ------------------------------------------------------------
class Student:
    #_marks= 95
    def __init__(self):
        self._marks = 95   # protected variable

    def show(self):

        print('Marks from Student Class: ',self._marks)


class Child(Student):
    def show(self):

        super().show()
        self._marks=100
        print('Marks from Clild Class: ',self._marks)

c = Child()
c.show()

# Explanation:
# A single underscore is a convention meaning:
# "Intended for internal use; do not access from outside".
# Not enforced — just a guideline.
```

```python


# ------------------------------------------------------------
# Example 3 — Private attribute (__var)
# ------------------------------------------------------------
class Bank:
    def __init__(self):
        self.__balance = 5000  # private variable

    def get_balance(self):
        return self.__balance

b = Bank()
print(b.get_balance())  # ERROR: Cannot access private variable

# Explanation:
# Double underscore triggers name-mangling:
# b._Bank__balance is how Python internally stores it.
# Prevents accidental overrides in subclasses.

```

```python

# ------------------------------------------------------------
# Example 4 — Accessing private variables using name-mangling
# ------------------------------------------------------------
class Bank:
    def __init__(self):
        self.__balance = 7000

b = Bank()
print(b._Bank__balance)

# Explanation:
# Private variables are still accessible if you know name-mangled format.
# This is NOT encouraged — used internally for safety.

```

```python

# ------------------------------------------------------------
# Example 5 — Getter and Setter for private attributes
# ------------------------------------------------------------
class Account:
    def __init__(self):
        self.__balance = 1000

    def get_balance(self):
        # controlled read access
        return self.__balance

    def set_balance(self, amt):
        # validation before updating
        if amt > 0:
            self.__balance = amt

a = Account()
a.set_balance(2000)
print(a.get_balance())

# Explanation:
# Encapsulation ensures validation and safety before updating data.
# Direct access is blocked; access is routed through safe methods.

```

```python


# ------------------------------------------------------------
# Example 6 — Private method (__method)
# ------------------------------------------------------------
class Car:
    def __init__(self):
        self.engine_status = "OFF"

    def __start_engine(self):
        # private helper method
        self.engine_status = "ON"

    def start(self):
        self.__start_engine()
        print("Engine Started. Status =", self.engine_status)

c = Car()
c.start()

# Explanation:
# Private methods hide internal steps from external users.
# Users interact only with high-level API methods (start()).
```

```python

# ------------------------------------------------------------
# Example 7 — Encapsulation through methods (data hiding)
# ------------------------------------------------------------
class Temperature:
    def __init__(self):
        self.__celsius = 25

    def get_fahrenheit(self):
        return (self.__celsius * 9/5) + 32

    def set_celsius(self, temp):
        if -273 <= temp:  # absolute zero check
            self.__celsius = temp

t = Temperature()
print(t.get_fahrenheit()) # Print the default celcius to farenhite converted value
t.set_celsius(30)
print(t.get_fahrenheit())
t.set_celsius(-40)
print(t.get_fahrenheit())


# Explanation:
# Internal data (__celsius) is hidden.
# Only safe operations allowed via methods.

```

```python

# ------------------------------------------------------------
# Example 8 — Enforcing valid data using encapsulation
# ------------------------------------------------------------
class Employee:
    def __init__(self):
        self.__salary = 30000

    def set_salary(self, amount):
        if amount < 0:
            print("Invalid Salary!")
        else:
            self.__salary = amount

    def get_salary(self):
        return self.__salary

e = Employee()
e.set_salary(50000)
print(e.get_salary())

# Explanation:
# Encapsulation helps prevent invalid data updates.


```

```python

# ------------------------------------------------------------
# Example 9 — Encapsulated behavior using protected variables
# ------------------------------------------------------------
class Vehicle:
    def __init__(self):
        self._speed = 0  # protected

    def accelerate(self, value):
        self._speed += value

class Bike(Vehicle):
    def show_speed(self):
        print("Current speed =", self._speed)

b = Bike()
b.accelerate(20)
b.show_speed()

# Explanation:
# Protected variable is meant to be used by subclasses.
# Not intended for direct public manipulation.
```

```python

# ------------------------------------------------------------
# Example 10 — Using properties (@property) for encapsulation
# ------------------------------------------------------------
class Product:

    def __init__(self, price):
        self.__price = price
        self.balance=1000

    @property
    def price(self):
        # getter using property decorator
        return self.__price

    @price.setter
    def price(self, value):
        # setter with validation
        if value > 0:
            self.__price = value

p = Product(100)
print(p.price)
p.price=100
print(p.price)
print(p.balance)

# Explanation:
# @property turns methods into attribute-like access.
# Encapsulation + clean syntax = preferred Pythonic approach.
```

```python
class Product:
    def __init__(self, price):
        self.__price = price


    def getPrice(self):
        # getter using property decorator
        return self.__price


    def setPrice(self, value):
        # setter with validation
        if value > 0:
            self.__price = value
```
